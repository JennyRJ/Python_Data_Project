```python
#Import libraries
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

#Cleaning the data
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df_europe =df[df["job_location"] == "Germany"]

df_europe.reset_index(inplace=True)
df_europe.head(10)
```

```python
df_europe.set_index("Job Id", inplace = True)

df_europe.head(5)
```

```python
Median_pivot = df_europe.pivot_table(index="job_title_short", values="salary_year_avg", aggfunc=["median", "max","min"])
Median_pivot
```

```python
Median_pivot.sort_values(by= ('median', 'salary_year_avg'),ascending =True)
```

```python
Median_pivot.sort_index(inplace=True)
```
