```python
#Import libraries
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

#Load the dataset
dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

#Clean the data
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df.pivot_table(index = "job_country", aggfunc ="median", values="salary_year_avg" ,columns= "job_title_short")
```

```python
top_countries =df["job_country"].value_counts().head(6).index
top_countries
```

```python
df_job_country_salary = df.pivot_table(
    index = "job_country",
    aggfunc = "median", 
    values = "salary_year_avg", 
    columns = "job_title_short")
df_job_country_salary = df_job_country_salary.loc[top_countries]

job_list = ["Data Analyst", "Data Engineer", "Data Scientist"]
df_job_country_salary = df_job_country_salary[job_list]
df_job_country_salary
```

```python
df_job_country_salary.plot(kind="bar")


plt.ylabel('Median salary per year')
plt.xlabel('')
plt.title('Median salary per year by country and job_title')
plt.xticks(rotation=45,ha = "right")


plt.show()
```
